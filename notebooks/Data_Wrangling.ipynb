# Data Wrangling – Medical Insurance Dataset

## Project purpose
This notebook prepares the raw medical insurance dataset for analysis and model development. It inspects the data, assigns meaningful column names, identifies missing-value markers, converts fields to appropriate types, handles missing values, and exports a cleaned CSV file.

**Workflow:** load raw data → inspect structure → name columns → identify missing values → clean selected fields → save the prepared dataset.

> Generative AI was used as a supporting learning and development aid to explain pandas operations, troubleshoot errors, and help document the workflow. Data decisions and code outputs should still be checked against the dataset.

```python
import pandas as pd
import numpy as np
```

## 1. Load the raw dataset

The CSV is read with `header=None` because this version of the source file is expected not to contain a header row. The first inspection helps confirm whether the file has been read as expected.

```python
df = pd.read_csv("medical_insurance_dataset.csv",header=None)
```

```python
print(df.head(10))
```

```python
df.info()
```

## 2. Assign column names

The dataset contains seven fields. Meaningful column names make later analysis and model code easier to read. Confirm that the order matches the source dataset before proceeding.

```python
df.columns = ["age", "gender", "bmi", "no_of_children", "smoker","region","charges"]
```

```python
print(df.head(10))
```

## 3. Identify missing-value markers

Some datasets represent missing values with a question mark (`?`) rather than a true null value. Replacing this marker with `NaN` allows pandas missing-value functions to detect it.

```python
df.replace('?', np.nan, inplace=True)
```

```python
print(df.head(10))
```

```python
df.info()
```

## 4. Inspect missing values and data types

The following checks help identify which columns contain missing observations and whether the columns have suitable data types for analysis.

```python
print(df.isnull().sum())
```

```python
df.info()
```

## 5. Convert age to numeric

`errors='coerce'` converts valid age values to numbers and turns values that cannot be interpreted as numbers into `NaN`. Those invalid values can then be handled as missing data.

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df.info()
```

## 6. Fill missing age values

This step fills missing ages using the mean age. The mean is one possible imputation choice; it can be affected by outliers. For a strongly skewed age distribution, the median may be more robust. Keep the chosen method consistent with the rationale in your project report.

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
print(df['age'].isnull().sum())
```

```python
print(df['age'].dtype)
print(df['age'].value_counts())
```

## 7. Handle missing smoker values

The mode is the most frequently occurring category. Here it is used to fill missing smoker entries. This assumes that using the most common category is an acceptable choice for this dataset.

```python
mode_value = df['smoker'].mode()[0]
```

```python
df['smoker'] = df['smoker'].fillna(mode_value)
```

```python
print(df['smoker'].isnull().sum())
print(df['smoker'].dtype)
print(df['smoker'].value_counts())
```

## 8. Review and export the prepared dataset

Inspect the cleaned records, save the result without writing the DataFrame index as an extra CSV column, and reload the exported file to verify that it can be read.

```python
print(df.head(10))
```

```python
df.to_csv("updated_medical_insurance_dataset.csv", index=False)
```
